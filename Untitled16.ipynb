```python
n = int(input())

sets = []

for _ in range(n):
    input()
    sets.append(set(input().split()))

freq = {}

for s in sets:
    for app in s:
        freq[app] = freq.get(app, 0) + 1

print("Hanya 1 kategori:")
for k in freq:
    if freq[k] == 1:
        print(k)

print("Tepat 2 kategori:")
for k in freq:
    if freq[k] == 2:
        print(k)
```

```python
list_data = [1, 2, 2, 3, 4]
set_data = {5, 6, 7}
tuple_data = (8, 9, 9, 10)

print("List sebelum :", list_data)
print("List -> Set   :", set(list_data))

print("\nSet sebelum  :", set_data)
print("Set -> List  :", list(set_data))

print("\nTuple sebelum :", tuple_data)
print("Tuple -> Set  :", set(tuple_data))

print("\nSet sebelum   :", set_data)
print("Set -> Tuple  :", tuple(set_data))
```

```python
try:
    file1 = input("Nama file 1: ")
    file2 = input("Nama file 2: ")

    with open(file1, "r") as f1:
        kata1 = set(f1.read().lower().split())

    with open(file2, "r") as f2:
        kata2 = set(f2.read().lower().split())

    sama = kata1.intersection(kata2)

    print("\nKata yang muncul di kedua file:")
    print(sama)

except FileNotFoundError:
    print("Error: File tidak ditemukan!")
except:
    print("Error: File tidak bisa dibaca!")
```
