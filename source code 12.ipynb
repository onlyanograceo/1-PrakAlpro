```python
isi_contoh = """From stephen.marquard@uct.ac.za Sat Jan  5
From louis@media.berkeley.edu Fri Jan  4
From zqian@umich.edu Fri Jan  4
From louis@media.berkeley.edu Fri Jan  4
"""
with open("mbox-short.txt", "w") as f:
    f.write(isi_contoh)

nama_file = input("Masukkan nama file : ")
counts = {}

try:
    hand = open(nama_file)
    for line in hand:
        if line.startswith("From "):
            words = line.split()
            email = words[1]
            counts[email] = counts.get(email, 0) + 1
    
    print(counts)
except:
    print("File tidak ditemukan!")
```

```python
fname = input("Masukkan nama file: ")

try:
    fhand = open(fname)
except:
    print("File tidak dapat dibuka:", fname)
    exit()

counts = dict()

for line in fhand:
    if line.startswith('From '):
        words = line.split()
        email = words[1]
        
        parts = email.split('@')
        domain = parts[1]
        
        counts[domain] = counts.get(domain, 0) + 1

print(counts)
```
