```python
def cek_prima(n, i=2):
    if n <= 1:
        return False
    if i == n:
        return True
    if n % i == 0:
        return False
    return cek_prima(n, i + 1)

angka = int(input("Masukkan bilangan: "))

if cek_prima(angka):
    print("Bilangan prima")
else:
    print("Bukan bilangan prima")
```

```python
def palindrome(kalimat):
    if len(kalimat) <= 1:
        return True
    if kalimat[0] != kalimat[-1]:
        return False
    return palindrome(kalimat[1:-1])

teks = input("Masukkan kalimat: ")

if palindrome(teks):
    print("Palindrome")
else:
    print("Bukan palindrome")
```

```python
def deret_ganjil(n):
    if n == 1:
        return 1
    return n + deret_ganjil(n - 2)

n = int(input("Masukkan bilangan ganjil terakhir: "))

print("Jumlah deret ganjil =", deret_ganjil(n))
```

```python
def jumlah_digit(n):
    n = abs(n)
    
    if n < 10:
        return n
    else:
        return (n % 10) + jumlah_digit(n // 10)

angka = 234
hasil = jumlah_digit(angka)
print(f"Jumlah digit dari {angka} adalah {hasil}")
```

```python
def kombinasi(n, k):
    if k == 0 or k == n:
        return 1
    else:
        return kombinasi(n - 1, k - 1) + kombinasi(n - 1, k)

n = 5
k = 2
hasil = kombinasi(n, k)
print(f"Kombinasi dari {n}C{k} adalah {hasil}")
```
