```python
fname = input("Enter a file name: ")
file = open(fname)

jam_dict = {}

for baris in file:
    kata = baris.split()
    if len(kata) >= 6 and kata[0] == "From":
        jam = kata[5].split(":")[0]
        jam_dict[jam] = jam_dict.get(jam, 0) + 1

for jam, jumlah in sorted(jam_dict.items()):
    print(jam, jumlah)
```
